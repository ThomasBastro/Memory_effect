```python
t = np.linspace(1e-3, 1e2, 5000)  # Time array in s
theta = np.deg2rad(30)  # Viewing angle in radians
M_ej = 0.05  # Ejecta mass in solar masses
v_pm_min=0.025
v_pm_max=0.15
v_d_min=0.15
v_max=0.9

h = u_em.compute_memory_from_density_profile(
    u_em.rho_profile_model1,
    t, theta, M_ej, v_pm_min, v_pm_max, v_d_min, v_max
)
plt.figure(figsize=(10, 6))
plt.plot(t, h, label='Memory from Density Profile', color='blue')
plt.xscale('log')
plt.yscale('log')
plt.xlabel('Time (s)')
plt.ylabel('Strain h')
plt.title('Memory from Ejecta Density Profile')
plt.grid()
plt.legend()
plt.show()
```
